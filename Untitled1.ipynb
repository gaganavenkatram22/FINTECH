```python
import re
import gradio as gr

# ---------------------------------------
# FinGuard AI
# Scam and Phishing Message Detector
# ---------------------------------------

SCAM_PATTERNS = {
    "Urgency or threats": [
        "account blocked",
        "account suspended",
        "act immediately",
        "urgent",
        "within 24 hours",
        "immediately"
    ],

    "Banking or KYC request": [
        "update kyc",
        "verify your bank account",
        "bank account blocked",
        "share your otp",
        "enter your otp",
        "bank details"
    ],

    "Prize or reward scam": [
        "you won",
        "winner",
        "claim your prize",
        "claim your reward",
        "cash prize",
        "free gift"
    ],

    "Financial scam": [
        "guaranteed returns",
        "double your money",
        "investment guaranteed",
        "send money to receive",
        "pay processing fee"
    ],

    "Credential theft": [
        "confirm your password",
        "verify your password",
        "login details",
        "share your password"
    ]
}


def analyze_message(message):

    if not message or not message.strip():
        return (
            "Please enter a message.",
            "No score",
            "No message analyzed.",
            "Enter a message and try again."
        )

    text = message.lower()
    score = 0
    findings = []

    # 1. Check known suspicious phrases
    for category, phrases in SCAM_PATTERNS.items():

        matched = [
            phrase for phrase in phrases
            if phrase in text
        ]

        if matched:
            score += min(25, len(matched) * 15)
            findings.append(
                category + ": " + ", ".join(matched)
            )

    # 2. Detect URLs
    urls = re.findall(
        r"https?://\S+|www\.\S+",
        text
    )

    if urls:
        score += 20
        findings.append(
            "Link detected: " + ", ".join(urls)
        )

        # Additional warning for URL shorteners
        shorteners = [
            "bit.ly",
            "tinyurl.com",
            "t.co",
            "is.gd",
            "cutt.ly"
        ]

        if any(
            shortener in url.lower()
            for url in urls
            for shortener in shorteners
        ):
            score += 15
            findings.append(
                "Shortened URL detected. "
                "Its final destination is not verified."
            )

    # 3. Detect requests for OTP or payment
    sensitive_requests = [
        "otp",
        "one time password",
        "send money",
        "transfer money",
        "pay now",
        "upi pin"
    ]

    matched_requests = [
        item for item in sensitive_requests
        if item in text
    ]

    if matched_requests:
        score += 20
        findings.append(
            "Sensitive request: "
            + ", ".join(matched_requests)
        )

    # Keep score between 0 and 100
    score = min(score, 100)

    # 4. Classify risk
    if score >= 60:
        status = "🚨 HIGH RISK"
        advice = (
            "Do not click suspicious links, send money, "
            "or share OTPs and passwords. Verify through "
            "your bank's official app or contact details."
        )

    elif score >= 30:
        status = "⚠️ MEDIUM RISK"
        advice = (
            "Check the sender and independently verify "
            "the request before taking any action."
        )

    else:
        status = "✅ LOW RISK INDICATORS"
        advice = (
            "No strong indicators were detected by these "
            "rules. This does not guarantee that the message "
            "is safe. Verify unexpected requests."
        )

    # 5. Show findings
    if findings:
        explanation = "\n".join(
            "- " + item for item in findings
        )
    else:
        explanation = (
            "No configured scam patterns or suspicious "
            "URL indicators were detected."
        )

    return (
        status,
        f"{score}/100",
        explanation,
        advice
    )


# ---------------------------------------
# User interface
# ---------------------------------------

with gr.Blocks(title="FinGuard AI") as app:

    gr.Markdown(
        """
        # 🛡️ FinGuard AI
        ### Real-Time Scam & Phishing Message Detector

        Paste a suspicious message below to analyze
        possible fraud indicators.

        This prototype analyzes text only. It does not
        connect to WhatsApp or verify URLs online.
        """
    )

    message_input = gr.Textbox(
        label="Enter WhatsApp Message",
        placeholder=(
            "Example: Your account will be blocked. "
            "Update KYC immediately at http://example.com"
        ),
        lines=5
    )

    check_button = gr.Button(
        "🔍 Check for Fraud",
        variant="primary"
    )

    status_output = gr.Textbox(
        label="Threat Classification"
    )

    score_output = gr.Textbox(
        label="Risk Score"
    )

    findings_output = gr.Textbox(
        label="Detected Indicators",
        lines=5
    )

    advice_output = gr.Textbox(
        label="Safety Recommendation",
        lines=3
    )

    check_button.click(
        fn=analyze_message,
        inputs=message_input,
        outputs=[
            status_output,
            score_output,
            findings_output,
            advice_output
        ]
    )

    gr.Examples(
        examples=[
            [
                "Congratulations! You won a cash prize. "
                "Claim your reward immediately."
            ],
            [
                "Your bank account will be blocked. "
                "Update KYC now at http://example.com"
            ],
            [
                "Hi, our college meeting is at 10 AM tomorrow."
            ]
        ],
        inputs=message_input
    )


app.launch(share=True)
```
